```python
import pandas as pd
import numpy as np
```

```python
df=pd.DataFrame(np.random.rand(10000,5),columns=('a','b','c','d','e'))
```

```python
df.describe()
```

```python
for i,row in df.iterrows():
    while i < 3:
        print(f'i={i}')
        print(' ')
        print('Returned Row')
        print(row)
        print(' ')
        print(f"Column B:{row['b']}")
        print("<----------------------------->")
        print(' ')
        break
        
```

```python
def iterrow_example(df, col):
    for i,  row in df.iterrows():
        val = row[col]
        if  val <.5:
            df.at[i,col]=0
        else:
            df.at[i,col]=1
                  
```

```python
%timeit  iterrow_example(df,'a')
```

```python
def iloc_example(df, col):
    for i in df.index:
        val = df[col].iloc[i]
        if  val <.5:
            df.at[i,col]=0
        else:
            df.at[i,col]=1
```

```python
%timeit  iterrow_example(df,'b')
```

```python
%timeit df['c']=df['c'].apply(lambda x : 0 if  x < .5 else 1)
```

```python
%timeit df['d'] = np.where(df['d'] < .5,0,1)
```

```python
%timeit df['d'] = np.where(df['d'].values < .5,0,1)
```
