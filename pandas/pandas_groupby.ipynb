```python
# transform
```

```python
dftransform=sns.load_dataset('tips')
```

```python
dftransform
```

```python
dftransform['medianbill'] = dftransform.groupby('time',observed =False)['total_bill'].transform('median')
```

```python
dftransform
```

```python
dftransform.query('time=="Lunch"')
```

```python
dftransform.query('time=="Dinner"')
```

```python
dftransform.query('total_bill > medianbill')
```

```python
# show a  lambda Example  -> z-score for each group. where data lies  on  normal distribution curve
```

```python
dftransform['zscore'] = dftransform.groupby('time',observed = False)['total_bill'].transform(lambda x : (x-x.mean())/x.std())
```

```python
dftransform
```

```python
# apply for lambda 
```

```python
dftransform.groupby('time',observed=False).apply(lambda x : (x['tip']/x['total_bill']).mean(),include_groups=False).to_frame('avgtipPercent').reset_index()
```

```python
dftransform.groupby('time',observed=False).apply(lambda x : ((x['tip']/x['total_bill'])*100).mean(),include_groups=False).to_frame('avgtipPercent').reset_index()
```
