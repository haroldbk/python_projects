```python
import pandas as pd
import numpy as np
```

```python
people={'first':['corey','jane','joe','chris',np.nan,None,'NA'],
       'last':['Schafer','Doe','Doe','Schafer',np.nan,np.nan,'Missing'],
        'email':['corey@email.com','jane@email.com','joe@email.com',None,np.nan,'anonymous@email.com','NA'],
         'age':['33','53','63','36',None,None,'Missing']}
```

```python
df = pd.DataFrame(people)
df.replace('NA',np.nan,inplace=True)
df.replace('Missing',np.nan,inplace=True)
```

```python
df
```

```python
df.dropna()
```

```python
df.dropna(axis='index',how='all',subset=['last','email'])
```

```python
df.dropna(axis='index', how='all')  #has  to be  all 
```
